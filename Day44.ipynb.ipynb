Based on the confusion matrix, I explored Precision and understood its importance in minimizing false positives.
I then learned about Recall, which focuses on identifying all relevant positive cases.
The concept of F1-Score was covered as a balance between precision and recall.
I analyzed scenarios where accuracy alone can be misleading for imbalanced datasets.
The session also included an introduction to the ROC Curve and its role in evaluating classifier performance.
I learned how the AUC (Area Under Curve) value indicates the model’s ability to distinguish between classes.
Practical examples were used to visualize ROC curves.
In addition, we studied regression evaluation metrics such as MAE, MSE, RMSE, and R² score.
I understood how these metrics quantify prediction error in regression models.
