Today, I worked on Dimensionality Reduction using PCA.
I learned that high-dimensional data often contains redundant information.
PCA helps reduce noise while preserving important patterns.
I started by standardizing the dataset before applying PCA.
Standardization ensured that all features contributed equally.
I used PCA implementation from scikit-learn.
I analyzed the explained variance of each principal component.
I created a Scree Plot to identify the optimal number of components.
Based on the plot, I reduced 13 features to 3 components.
This retained nearly 96% of the total variance.
I understood that variance represents useful information in data.
PCA creates new features instead of selecting existing ones.
These new features are uncorrelated with each other.
I learned about orthogonality between principal components.
I visualized the transformed data using a 2D scatter plot.
The visualization clearly showed underlying data clusters.
I observed that excessive reduction can cause loss of detail.
PCA improves model efficiency but reduces human interpretability.
This task improved my understanding of linear algebra concepts.
Overall, the activity enhanced my data preprocessing skills.
